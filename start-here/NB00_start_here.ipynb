## 1. Variables and units

A variable is a name for a value. Python does not know units, so good names and comments carry them. The stress in a bar is force divided by area.

```python
force_kN = 25.0                 # axial force in kilonewtons
diameter_mm = 16.0              # bar diameter in millimetres
area_mm2 = 3.14159 * diameter_mm ** 2 / 4       # ** means "to the power of"
stress_MPa = force_kN * 1000 / area_mm2          # N / mm2 is the same as MPa
print(f"area {area_mm2:.1f} mm2, stress {stress_MPa:.1f} MPa")
```

The `f` before the quotation marks makes an f-string: Expressions in braces are inserted, and `:.1f` prints one decimal.

## 2. Lists and loops

A list holds several values in order. A `for` loop repeats the indented lines once for every element. The loads below act on the floors of a small building.

```python
floor_loads_kN = [120.0, 118.5, 121.2, 119.8, 95.0]
total = 0.0
for load in floor_loads_kN:
    total = total + load
print("number of floors:", len(floor_loads_kN))
print("total load:", total, "kN")
print("largest:", max(floor_loads_kN), "smallest:", min(floor_loads_kN))
print("first floor:", floor_loads_kN[0], "top floor:", floor_loads_kN[-1])   # counting starts at 0
```

### Your turn, exercise 2

Compute the average floor load with a loop or with `sum` and `len`, and store it in `average_load`.

```python
#@title Reference solution (run this cell; open it only after trying)
def _average_load_reference():
    return sum(floor_loads_kN) / len(floor_loads_kN)
```

```python
average_load = None
check("Exercise 2", average_load, _average_load_reference(), tol=1e-9)
```

## 3. Decisions with if

An `if` statement runs a block only when a condition is true. Comparisons such as `<`, `>=` and `==` produce True or False.

```python
allowable_MPa = 160.0
for load in [20.0, 30.0, 45.0]:
    s = load * 1000 / area_mm2
    if s <= allowable_MPa:
        verdict = "OK"
    else:
        verdict = "too high"
    print(f"{load:5.1f} kN -> {s:6.1f} MPa: {verdict}")
```

## 4. Functions

A function packages a calculation so that it can be reused. `def` defines it, the names in parentheses are its inputs, and `return` gives back the result. The tip deflection of a cantilever with a point load is P L^3 / (3 E I) [1].

```python
def cantilever_deflection(P, L, E, I):
    """Tip deflection of a cantilever with an end load P (all in SI units)."""
    return P * L ** 3 / (3 * E * I)


b, h = 0.05, 0.10                        # rectangular section in metres
I = b * h ** 3 / 12                      # second moment of area
d = cantilever_deflection(P=2000, L=1.5, E=200e9, I=I)
print(f"tip deflection {d * 1000:.2f} mm")
```

### Your turn, exercise 3

## 5. Dictionaries

A dictionary maps names to values, like a small table. Here each material maps to its elastic modulus and density.

```python
materials = {"steel": {"E": 200e9, "rho": 7850}, "aluminium": {"E": 70e9, "rho": 2700}, "timber": {"E": 11e9, "rho": 500}}
for name, props in materials.items():
    d = cantilever_deflection(2000, 1.5, props["E"], I)
    mass = props["rho"] * b * h * 1.5
    print(f"{name:10s} deflection {d * 1000:7.2f} mm, mass {mass:5.1f} kg")
```

## 6. A first plot

Matplotlib draws charts. `plt.plot(x, y)` connects points with lines. The chart shows an idealised stress-strain curve of steel: elastic up to the yield strength, then perfectly plastic.

```python
import matplotlib.pyplot as plt

E_steel, fy = 200e3, 355.0                # MPa
strains = [i * 0.0001 for i in range(0, 60)]
stresses = [min(E_steel * e, fy) for e in strains]
plt.figure(figsize=(6, 3.4))
plt.plot(strains, stresses)
plt.xlabel("strain"); plt.ylabel("stress (MPa)"); plt.title("Elastic-perfectly plastic steel")
plt.show()
```

## 7. A glimpse of NumPy
