## Collective behaviour as computation

A flock of birds turns as one body although no bird leads it. Reynolds reproduced this with three local rules for each simulated bird: avoid crowding neighbours, align with their heading, and move towards their centre [8]. Ant colonies find short paths between nest and food without a map, because ants deposit pheromone and prefer trails with more of it; shorter trails are completed faster and accumulate pheromone sooner. Honeybee colonies allocate foragers to flower patches in proportion to their quality through the waggle dance. These systems share decentralised control, simple agents, local interaction and emergent global behaviour, which Bonabeau, Dorigo and Theraulaz summarised as swarm intelligence [1].

## Particle swarm optimisation

Kennedy and Eberhart introduced particle swarm optimisation in 1995, inspired by bird flocking and social behaviour [2]. Each particle i has a position x, which is a candidate solution, and a velocity v. It remembers the best position p it has visited, and the swarm shares the best position g found by any particle. In every iteration, the velocity is updated as v = w v + c1 r1 (p - x) + c2 r2 (g - x), and the position as x = x + v, where r1 and r2 are random numbers between 0 and 1 drawn for each dimension.

The three terms have clear meanings. The inertia weight w keeps part of the previous motion; Shi and Eberhart introduced it to balance global and local search, with values around 0.4 to 0.9 [9]. The cognitive term, weighted by c1, pulls a particle back towards its own best experience. The social term, weighted by c2, pulls it towards the best experience of the swarm. Clerc and Kennedy analysed the dynamics and derived a constriction factor that guarantees convergence; the widely used setting w = 0.7298 with c1 = c2 = 1.49618 follows from it [10]. Particle swarms are easy to implement, need no gradients, and work well on continuous problems of moderate dimension, but like every metaheuristic they can stagnate in a local optimum.

## Tuning a PID controller with a swarm

The proportional-integral-derivative controller is the workhorse of industrial control. Its three gains shape the response: The proportional gain speeds up the reaction, the integral gain removes steady-state error, and the derivative gain damps oscillation [11]. Tuning them by hand is tedious, and classical rules assume simple process models. Controller tuning can instead be formulated as optimisation: Choose the gains that minimise a cost computed from a simulated step response, such as the integral of the time-weighted absolute error (ITAE), with penalties for overshoot and limits on the actuator. Particle swarms have been used in this way for automatic voltage regulators and many other loops [12].

The notebook applies the method to the speed control of a DC motor, using the standard model and parameters of the Control Tutorials for MATLAB and Simulink [13]. The armature voltage is limited to 24 volts, which makes the problem nonlinear, and the swarm evaluates all particles in one vectorised simulation. The cost function is part of the design: A cost that rewards speed alone drives the gains to their bounds, and adding a penalty on overshoot or control effort changes the optimum. The optimiser serves the engineer's definition of good behaviour; it cannot supply that definition.

## Ant colony optimisation

Dorigo, Maniezzo and Colorni turned the pheromone mechanism into the Ant System for the travelling salesman problem [3]. Each artificial ant builds a complete tour city by city. From city i, it chooses the next unvisited city j with a probability proportional to the pheromone on edge (i, j) raised to a power alpha, times a heuristic desirability, usually the inverse distance, raised to a power beta. After all ants have finished, pheromone evaporates on every edge by a factor rho, and each ant deposits pheromone on the edges of its tour in inverse proportion to the tour length. Short tours therefore reinforce their edges, while evaporation forgets poor early choices.

The Ant Colony System added a stronger exploitation rule and local pheromone updates and was competitive on benchmark instances [14]; the monograph by Dorigo and Stützle presents the family and its applications to routing, scheduling and assignment [4]. Drilling holes in a printed circuit board is a classical travelling salesman application: The drill head must visit every hole once, and the travel time between holes is wasted production time. Benchmark libraries such as TSPLIB include instances that come from drilling problems [15].

## Artificial bee colony and the metaphor question

Karaboga and Basturk's artificial bee colony algorithm assigns roles to solutions [5]. Employed bees search around their food sources, onlooker bees choose sources in proportion to their quality and search around them, and scout bees replace sources that have not improved for a number of trials with random new ones. The scout phase gives the algorithm a built-in restart mechanism. Many further algorithms followed, inspired by fireflies, bats, wolves, whales and others, including the ant lion optimiser used to train neural networks for chaotic signal prediction in earlier work [16].

Sörensen argued that the flood of metaphor-based algorithms often hides old ideas behind new vocabulary and weak experiments [6]. The no free lunch theorems add that no algorithm can be best on all problems [17]. Engineers should therefore judge a metaheuristic by its search mechanism rather than its metaphor, compare it with established methods on the problem at hand, report results over many independent runs with statistics, and count function evaluations, not iterations, because a single evaluation of an engineering simulation may take minutes.

### Why classes

Simulation code keeps state that changes over time, such as the speed and the current of a motor, together with the functions that change it. A class bundles both: Attributes store the state, and methods are functions that act on it. The class is the blueprint, and every object created from it has its own attributes. The special method `__init__` runs when an object is created and sets its initial state, and the first parameter of every method, `self`, refers to the object itself.

The DC motor of this week has the parameters of the speed example of the Control Tutorials for MATLAB and Simulink [13]. Its speed w and current i obey J dw/dt = K i - b w and L di/dt = V - R i - K w, which the method `step` integrates over small time steps.

```python
class DCMotor:
    """Armature-controlled DC motor with the parameters of the CTMS speed example."""

    def __init__(self, J=0.01, b=0.1, K=0.01, R=1.0, L=0.5):
        self.J, self.b, self.K, self.R, self.L = J, b, K, R, L
        self.speed = 0.0        # rad/s
        self.current = 0.0      # A

    def step(self, voltage, dt=0.001):
        """Advance the motor by dt seconds with the explicit Euler method."""
        acc = (self.K * self.current - self.b * self.speed) / self.J
        di = (voltage - self.R * self.current - self.K * self.speed) / self.L
        self.speed += acc * dt
        self.current += di * dt
        return self.speed


motor = DCMotor()
for _ in range(3000):           # three seconds in steps of 1 ms
    motor.step(1.0)
print(f"speed after 3 s at 1 V: {motor.speed:.4f} rad/s")
```

The underscore `_` is the conventional name for a loop variable that is not used. The speed approaches K V / (b R + K^2), about 0.1 rad/s for 1 V, which the steady-state analysis of the tutorial also gives.

### Many objects from one class

Objects created from the same class are independent. A motor with a lighter rotor does not share its speed with a motor with a heavier one, and each object keeps its own attributes. This independence makes classes practical for swarms, fleets of vehicles and banks of sensors.

```python
light = DCMotor(J=0.005)
heavy = DCMotor(J=0.05)
for _ in range(200):            # 0.2 s
    light.step(1.0)
    heavy.step(1.0)
print(f"light rotor: {light.speed:.4f} rad/s, heavy rotor: {heavy.speed:.4f} rad/s")
```

### A particle of the swarm

Particle swarm optimisation moves each particle with a velocity that combines its previous velocity, a pull towards its own best position and a pull towards the best position of the swarm. A class keeps the position, the velocity and the personal best of a particle together, and NumPy arrays let the same code work in any number of dimensions. The method `copy` creates an independent array: Without it, the personal best and the current position would be two names for the same array, and changing one in place would also change the other.

```python
import numpy as np


class Particle:
    """One particle of a swarm with position, velocity and personal best."""

    def __init__(self, position, velocity):
        self.x = np.array(position, dtype=float)
        self.v = np.array(velocity, dtype=float)
        self.best_x = self.x.copy()

    def move(self, swarm_best, w=0.7, c1=1.5, c2=1.5, r1=0.5, r2=0.5):
        """Update velocity and position; r1 and r2 stand in for random numbers."""
        self.v = w * self.v + c1 * r1 * (self.best_x - self.x) + c2 * r2 * (swarm_best - self.x)
        self.x = self.x + self.v


p = Particle([2.0, -1.0], [0.1, 0.0])
p.move(swarm_best=np.array([0.0, 0.0]))
print(p.x, p.v)
```

The notebook of this week builds a complete swarm on the same pattern, with new random numbers at every step.

### Probabilistic choice in an ant colony

Ant colony optimisation builds solutions step by step [3]. At every step an ant chooses among the possible next moves with probabilities proportional to tau^alpha eta^beta, where tau is the pheromone on a move and eta its heuristic desirability, usually one over the distance. The generator method `choice` draws from such a discrete distribution when the probabilities are passed as `p`, and `np.bincount` counts how often each move was chosen.

```python
tau = np.ones(4)                                # pheromone on four candidate edges
dist = np.array([4.0, 2.0, 5.0, 3.0])           # edge lengths in km
alpha, beta = 1.0, 2.0
weight = tau ** alpha * (1 / dist) ** beta
prob = weight / weight.sum()
print(prob.round(3))
colony = np.random.default_rng(5)
choices = colony.choice(4, size=1000, p=prob)   # the choices of 1000 ants
print(np.bincount(choices, minlength=4))
```

After the ants have moved, pheromone evaporates at a rate rho, and every ant deposits pheromone on its choice, more on shorter edges. Repeated over many iterations, this positive feedback concentrates the colony on short routes, while evaporation keeps it from settling too early on a poor one.

```python
rho = 0.5
deposit = np.bincount(choices, minlength=4) / dist / len(choices)   # 1/length per ant, scaled by the colony size
tau = (1 - rho) * tau + deposit
print(tau.round(3))
```

## 2. A particle swarm as a class

**Python move: classes.** `PSO` stores the swarm state as attributes and offers `step` and `run` methods. The objective receives all positions at once and returns one cost per particle, so vectorised objectives run fast.

```python
class PSO:
    def __init__(self, cost, lo, hi, n=30, w=0.7298, c1=1.49618, c2=1.49618, seed=0):
        self.cost, self.lo, self.hi = cost, np.asarray(lo, float), np.asarray(hi, float)
        self.w, self.c1, self.c2 = w, c1, c2
        self.rng = np.random.default_rng(seed)
        self.x = self.lo + (self.hi - self.lo) * self.rng.random((n, len(self.lo)))
        self.v = 0.1 * (self.hi - self.lo) * self.rng.uniform(-1, 1, self.x.shape)
        self.p, self.pf = self.x.copy(), self.cost(self.x)
        self.g = self.p[np.argmin(self.pf)].copy()
        self.history = [self.pf.min()]
        self.evaluations = n

    def step(self):
        r1, r2 = self.rng.random(self.x.shape), self.rng.random(self.x.shape)
        self.v = self.w * self.v + self.c1 * r1 * (self.p - self.x) + self.c2 * r2 * (self.g - self.x)
        self.x = np.clip(self.x + self.v, self.lo, self.hi)
        f = self.cost(self.x)
        self.evaluations += len(f)
        improved = f < self.pf
        self.p[improved], self.pf[improved] = self.x[improved], f[improved]
        self.g = self.p[np.argmin(self.pf)].copy()
        self.history.append(self.pf.min())

    def run(self, iterations=100):
        for _ in range(iterations):
            self.step()
        return self.g, self.pf.min()


def rastrigin(X):
    X = np.atleast_2d(X)
    return 10 * X.shape[1] + np.sum(X**2 - 10 * np.cos(2 * np.pi * X), axis=1)


for seed in range(3):
    best_x, best_f = PSO(rastrigin, [-5.12] * 5, [5.12] * 5, n=40, seed=seed).run(200)
    print(f"seed {seed}: best value {best_f:.4f}")
```

## 3. The DC motor speed loop

The motor model and its parameters come from the Control Tutorials for MATLAB and Simulink: J = 0.01 kg m2, b = 0.1 N m s, K = 0.01, R = 1 ohm, L = 0.5 H [13]. Here the armature voltage is limited to 24 V and the derivative term is filtered with a 0.01 s time constant. `simulate_pid` simulates many gain sets at once: each row of `gains` is one particle.

**Python move: vectorised simulation.** The time loop runs once, and every line inside it updates all particles together.

```python
J, B_FR, K_M, R_A, L_A, V_MAX = 0.01, 0.1, 0.01, 1.0, 0.5, 24.0
DT, T_END = 0.002, 2.0


def simulate_pid(gains):
    gains = np.atleast_2d(gains)
    kp, ki, kd = gains[:, 0], gains[:, 1], gains[:, 2]
    n = len(kp)
    w, i, integ, d_f = (np.zeros(n) for _ in range(4))
    e_prev = np.ones(n)
    steps = int(T_END / DT)
    trace = np.zeros((steps, n))
    for k in range(steps):
        e = 1.0 - w                                   # unit step in speed (rad/s)
        integ += e * DT
        d_f += ((e - e_prev) / DT - d_f) * DT / 0.01  # filtered derivative
        e_prev = e
        volts = np.clip(kp * e + ki * integ + kd * d_f, -V_MAX, V_MAX)
        w, i = w + (K_M * i - B_FR * w) / J * DT, i + (volts - R_A * i - K_M * w) / L_A * DT
        trace[k] = w
    return trace


t = np.arange(int(T_END / DT)) * DT


def pid_cost(gains, overshoot_weight=20.0):
    trace = simulate_pid(gains)
    itae = np.sum(t[:, None] * np.abs(1.0 - trace), axis=0) * DT
    overshoot = np.maximum(0.0, trace.max(axis=0) - 1.02)
    return itae + overshoot_weight * overshoot


reference = np.array([100.0, 200.0, 10.0])        # tuning used in the Control Tutorials PID example
print("cost of the reference gains:", pid_cost(reference)[0].round(4))
```

The reference gains were designed for the linear model. With the 24 V limit, the voltage saturates at the start of the step, the integral term keeps growing during saturation, and the stored integral causes overshoot later. This effect is called integral windup [11].

## 4. Tuning with the swarm

The search bounds are 0 to 300 for Kp, 0 to 600 for Ki and 0 to 30 for Kd. Every iteration simulates all particles at once.

## 5. An ant colony for a drilling sequence

Thirty holes on a board must be drilled in one closed tour. The Ant System chooses the next hole with probability proportional to pheromone^alpha (1 / distance)^beta, evaporates pheromone by rho and deposits 1 / length on each ant's tour [3, 4].

```python
rng = np.random.default_rng(11)
centres = np.array([[0.25, 0.3], [0.7, 0.35], [0.5, 0.75]])
holes = np.clip(centres[rng.integers(0, 3, 30)] + 0.13 * rng.normal(size=(30, 2)), 0.02, 0.98)
D = np.linalg.norm(holes[:, None, :] - holes[None, :, :], axis=2) + np.eye(30)


def tour_length(tour):
    return float(sum(D[tour[k], tour[(k + 1) % len(tour)]] for k in range(len(tour))))


def nearest_neighbour(start=0):
    tour, left = [start], set(range(len(D))) - {start}
    while left:
        nxt = min(left, key=lambda j: D[tour[-1], j])
        tour.append(nxt); left.remove(nxt)
    return tour


def ant_system(n_ants=20, iterations=150, alpha=1.0, beta=3.0, rho=0.3, seed=0):
    rng = np.random.default_rng(seed)
    n = len(D)
    tau, eta = np.ones((n, n)), 1.0 / D
    best, best_len, history = None, np.inf, []
    for _ in range(iterations):
        tours = []
        for _ in range(n_ants):
            tour = [rng.integers(n)]
            visited = np.zeros(n, bool); visited[tour[0]] = True
            for _ in range(n - 1):
                weights = (tau[tour[-1]] ** alpha) * (eta[tour[-1]] ** beta) * (~visited)
                nxt = rng.choice(n, p=weights / weights.sum())
                tour.append(nxt); visited[nxt] = True
            tours.append(tour)
        tau *= (1 - rho)                                        # evaporation
        for tour in tours:
            L = tour_length(tour)
            if L < best_len:
                best, best_len = tour, L
            for k in range(n):
                a, b = tour[k], tour[(k + 1) % n]
                tau[a, b] += 1 / L; tau[b, a] += 1 / L           # deposit
        history.append(best_len)
    return best, best_len, history


nn = nearest_neighbour()
best_tour, best_len, hist = ant_system()
print(f"nearest neighbour {tour_length(nn):.3f}   ant system {best_len:.3f}")
fig, ax = plt.subplots(1, 2, figsize=(9, 3.6))
for a, tour, title in [(ax[0], nn, "nearest neighbour"), (ax[1], best_tour, "ant system")]:
    loop_ = holes[list(tour) + [tour[0]]]
    a.plot(loop_[:, 0], loop_[:, 1], "-o", ms=3); a.set_title(f"{title}: {tour_length(tour):.3f}"); a.set_aspect("equal"); a.set_xticks([]); a.set_yticks([])
plt.show()
```
